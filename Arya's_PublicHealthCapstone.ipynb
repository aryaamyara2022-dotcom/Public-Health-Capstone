```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
import pandas as pd

url = "https://data.cdc.gov/api/views/d2rk-yvas/rows.csv?accessType=DOWNLOAD"
df = pd.read_csv(url)

# Filter data for 'Cardiovascular Disease' and specified years (2011-2013)
filtered_cardiovascular_data = df[
    (df['Topic'] == "Cardiovascular Disease") &
    (df['Year'].isin([2011, 2012, 2013]))
].copy()

# Drop rows where 'Data_value' is null, as this is our target variable
filtered_cardiovascular_data.dropna(subset=['Data_value'], inplace=True)

# Define features (X) and target (y)
# Use 'Year' and 'Sample_Size' as numerical features
# Use 'Locationdesc', 'Class', 'Question', 'Response' as categorical features
features = ['Year', 'Sample_Size', 'Locationdesc', 'Class', 'Question', 'Response']
X_cardio = filtered_cardiovascular_data[features]
y_cardio = filtered_cardiovascular_data['Data_value']

# Convert categorical features to numerical using one-hot encoding
X_cardio = pd.get_dummies(X_cardio, columns=['Locationdesc', 'Class', 'Question', 'Response'], drop_first=True)

# Split the data into training and testing sets
X_cardio_train, X_cardio_test, y_cardio_train, y_cardio_test = train_test_split(X_cardio, y_cardio, test_size=0.2, random_state=42)

print(f"Shape of training features for Cardiovascular Disease: {X_cardio_train.shape}")
print(f"Shape of testing features for Cardiovascular Disease: {X_cardio_test.shape}")
print(f"Shape of training target for Cardiovascular Disease: {y_cardio_train.shape}")
print(f"Shape of testing target for Cardiovascular Disease: {y_cardio_test.shape}")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

# Initialize and train the Linear Regression model for Cardiovascular Disease
linear_model_cardio = LinearRegression()
linear_model_cardio.fit(X_cardio_train, y_cardio_train)

# Make predictions on the test set for Cardiovascular Disease
y_cardio_pred = linear_model_cardio.predict(X_cardio_test)

# Evaluate the model
mse_cardio = mean_squared_error(y_cardio_test, y_cardio_pred)
r2_cardio = r2_score(y_cardio_test, y_cardio_pred)

print(f"Linear Regression Model Evaluation for Cardiovascular Disease (2011-2013):")
print(f"Mean Squared Error (MSE): {mse_cardio:.4f}")
print(f"R-squared (R2): {r2_cardio:.4f}")

# Visualize actual vs. predicted values for Cardiovascular Disease
fig = plt.figure(figsize=(10, 6))
sns.scatterplot(x=y_cardio_test, y=y_cardio_pred, alpha=0.6)
plt.plot([y_cardio.min(), y_cardio.max()], [y_cardio.min(), y_cardio.max()], 'r--', lw=2) # Diagonal line for perfect prediction
plt.xlabel("Actual Data_value")
plt.ylabel("Predicted Data_value")
plt.title("Linear Regression: Actual vs. Predicted Data_value (Cardiovascular Disease 2011-2013)")
plt.grid(True)
plt.show()
```

**I observed that as the predicated data values increase, the number of actual data values increase. Not only that but, the line chart is getting better over time. All the data points cluster very closely around the red dashed line.**

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import mean_squared_error, r2_score
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Load the dataset to ensure 'df' is defined in this execution context
url = "https://data.cdc.gov/api/views/d2rk-yvas/rows.csv?accessType=DOWNLOAD"
df = pd.read_csv(url)

# Filter data for 'Asthma' topic and specified years (2011-2013)
filtered_asthma_data_model = df[
    (df['Topic'] == 'Asthma') &
    (df['Year'].isin([2011, 2012, 2013]))
].copy()

# Drop rows where 'Data_value' is null, as this is our target variable
filtered_asthma_data_model.dropna(subset=['Data_value'], inplace=True)

# Define features (X) and target (y)
features_gb_asthma = ['Year', 'Sample_Size', 'Locationdesc', 'Class', 'Question', 'Response', 'Break_Out', 'Break_Out_Category']
X_gb_asthma = filtered_asthma_data_model[features_gb_asthma]
y_gb_asthma = filtered_asthma_data_model['Data_value']

# Convert categorical features to numerical using one-hot encoding
X_gb_asthma = pd.get_dummies(X_gb_asthma, columns=['Locationdesc', 'Class', 'Question', 'Response', 'Break_Out', 'Break_Out_Category'], drop_first=True, dtype=int)

# Split the data into training and testing sets
X_gb_asthma_train, X_gb_asthma_test, y_gb_asthma_train, y_gb_asthma_test = train_test_split(X_gb_asthma, y_gb_asthma, test_size=0.2, random_state=42)

print(f"Shape of training features for Asthma: {X_gb_asthma_train.shape}")
print(f"Shape of testing features for Asthma: {X_gb_asthma_test.shape}")
print(f"Shape of training target for Asthma: {y_gb_asthma_train.shape}")
print(f"Shape of testing target for Asthma: {y_gb_asthma_test.shape}")

# Initialize and train the Gradient Boosting Regressor model
gradient_boost_asthma_model = GradientBoostingRegressor(n_estimators=100, learning_rate=0.1, max_depth=3, random_state=42)
gradient_boost_asthma_model.fit(X_gb_asthma_train, y_gb_asthma_train)

# Make predictions on the test set
y_gb_asthma_pred = gradient_boost_asthma_model.predict(X_gb_asthma_test)

# Evaluate the model
mse_gb_asthma = mean_squared_error(y_gb_asthma_test, y_gb_asthma_pred)
r2_gb_asthma = r2_score(y_gb_asthma_test, y_gb_asthma_pred)

print(f"\nGradient Boosting Regressor Model Evaluation (Asthma 2011-2013):")
print(f"Mean Squared Error (MSE): {mse_gb_asthma:.4f}")
print(f"R-squared (R2): {r2_gb_asthma:.4f}")

# Add jitter to predicted values to make them more spread out for visualization
jitter_amount = 0.5 # Adjust this value as needed
y_gb_asthma_pred_jittered = y_gb_asthma_pred + np.random.normal(0, jitter_amount, y_gb_asthma_pred.shape)

# Visualize actual vs. predicted values with jitter
fig = plt.figure(figsize=(10, 6))
sns.scatterplot(x=y_gb_asthma_test, y=y_gb_asthma_pred_jittered, alpha=0.6) # Use jittered predictions
plt.plot([y_gb_asthma.min(), y_gb_asthma.max()], [y_gb_asthma.min(), y_gb_asthma.max()], 'r--', lw=2) # Diagonal line for perfect prediction
plt.xlabel("Actual Data_value")
plt.ylabel("Predicted Data_value (with jitter)") # Update label to reflect jitter
plt.title("Gradient Boosting Regressor: Actual vs. Predicted Data_value (Asthma 2011-2013) with Jitter") # Update title
plt.grid(True)
plt.show()
```

```python
import pandas as pd
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Normalization
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset again to ensure 'df' is defined in this execution context
url = "https://data.cdc.gov/api/views/d2rk-yvas/rows.csv?accessType=DOWNLOAD"
df = pd.read_csv(url)

# Filter data for 'Physical Activity' and specified years (2011-2013)
filtered_physical_activity_data = df[
    (df['Class'] == 'Physical Activity') &
    (df['Year'].isin([2011, 2012, 2013]))
].copy()

# Drop rows where 'Data_value' is null, as this is our target variable
filtered_physical_activity_data.dropna(subset=['Data_value'], inplace=True)

# Define features (X) and target (y)
features_nn = ['Year', 'Sample_Size', 'Locationdesc', 'Topic', 'Question', 'Response', 'Break_Out', 'Break_Out_Category']
X_nn = filtered_physical_activity_data[features_nn]
y_nn = filtered_physical_activity_data['Data_value']

# Convert categorical features to numerical using one-hot encoding, ensuring integer output
X_nn = pd.get_dummies(X_nn, columns=['Locationdesc', 'Topic', 'Question', 'Response', 'Break_Out', 'Break_Out_Category'], drop_first=True, dtype=int)

# Split the data into training and testing sets
X_nn_train, X_nn_test, y_nn_train, y_nn_test = train_test_split(X_nn, y_nn, test_size=0.2, random_state=42)

print(f"Shape of training features for Physical Activity: {X_nn_train.shape}")
print(f"Shape of testing features for Physical Activity: {X_nn_test.shape}")
print(f"Shape of training target for Physical Activity: {y_nn_train.shape}")
print(f"Shape of testing target for Physical Activity: {y_nn_test.shape}")

# Convert DataFrames to NumPy arrays for TensorFlow compatibility
X_nn_train_np = X_nn_train.to_numpy()
X_nn_test_np = X_nn_test.to_numpy()

# Normalize numerical features
normalizer = Normalization(axis=-1)
normalizer.adapt(X_nn_train_np)

# Build the Neural Network model
model = Sequential([
    normalizer,
    Dense(64, activation='relu'),
    Dense(32, activation='relu'),
    Dense(1) # Output layer for regression
])

model.compile(optimizer='adam', loss='mean_squared_error')

# Train the model
history = model.fit(X_nn_train_np, y_nn_train, epochs=50, validation_split=0.2, verbose=0)

# Make predictions on the test set
y_nn_pred = model.predict(X_nn_test_np).flatten()

# Evaluate the model
mse_nn = mean_squared_error(y_nn_test, y_nn_pred)
r2_nn = r2_score(y_nn_test, y_nn_pred)

print(f"\nNeural Network Model Evaluation (Physical Activity 2011-2013):")
print(f"Mean Squared Error (MSE): {mse_nn:.4f}")
print(f"R-squared (R2): {r2_nn:.4f}")

# Visualize actual vs. predicted values
fig = plt.figure(figsize=(10, 6))
sns.scatterplot(x=y_nn_test, y=y_nn_pred, alpha=0.6)
plt.plot([y_nn.min(), y_nn.max()], [y_nn.min(), y_nn.max()], 'r--', lw=2) # Diagonal line for perfect prediction
plt.xlabel("Actual Data_value")
plt.ylabel("Predicted Data_value")
plt.title("Neural Network: Actual vs. Predicted Data_value (Physical Activity 2011-2013)")
plt.grid(True)
plt.show()
```
